1.1.4 Написать функцию, которая для целых значений `n` и `m` будет возвращать тензор размера `n`x`m`, заполненный текстурой размера 2x2, состоящей из следующих значений:

```
0 1

2 3

```

Пример для n = 4 и m = 5:

```
0 1 0 1 0

2 3 2 3 2

0 1 0 1 0

2 3 2 3 2
```

````python
def texture(n, m):
    pattern = torch.tensor([[0, 1], [2, 3]])
    return pattern.repeat((n + 1) // 2, (m + 1) // 2)[:n, :m]

texture(4, 5)
````

1.3.14 Создать тензор `arr` из 20 случайных целых чисел от 0 до 100. Найти самое частое значение в тензоре. 
Найти индексы в тензоре, соответствующие самому частому значению. Проверить, как работет алгоритм при двух значениях, имеющих наибольшую встречаемость, предложить приемлемое поведение алгоритма для этого случая. 

````python
arr = torch.randint(0, 101, (20,))
counts = torch.bincount(arr)
max_count = counts.max()
modes = torch.nonzero(counts == max_count, as_tuple=True)[0]

mode = modes.min()
indices = torch.nonzero(arr == mode, as_tuple=True)[0]

print(arr)
print("Самые частые значения:", modes)
print("Выбранное значение:", mode.item())
print("Индексы:", indices)

arr_tie = torch.tensor([4, 4, 7, 7, 1, 2])
counts_tie = torch.bincount(arr_tie)
modes_tie = torch.nonzero(counts_tie == counts_tie.max(), as_tuple=True)[0]
print("Пример с двумя модами:", modes_tie)
print("Выбираем наименьшую:", modes_tie.min().item())
````

1.4.5 Найти "локальные максимумы" в одномерном тензоре (т.е. значения, большие предыдущего и последующего) `torch.tensor([1, 3, 7, 1, 2, 6, 0, 1])` и вывести их индексы.

````python
a = torch.tensor([1, 3, 7, 1, 2, 6, 0, 1])
mask = (a[1:-1] > a[:-2]) & (a[1:-1] > a[2:])
indices = torch.nonzero(mask, as_tuple=True)[0] + 1

print("Индексы локальных максимумов:", indices)
print("Локальные максимумы:", a[indices])
````

1.4.7 Решить матричное уравнение `A*X*B=-C` - найти матрицу X. Где `A = [[-1, 2, 4], [-3, 1, 2], [-3, 0, 1]]`, `B=[[3, -1], [2, 1]]`, `C=[[7, 21], [11, 8], [8, 4]]`.

````python
A = torch.tensor([[-1., 2., 4.],
                  [-3., 1., 2.],
                  [-3., 0., 1.]])
B = torch.tensor([[3., -1.],
                  [2.,  1.]])
C = torch.tensor([[7., 21.],
                  [11., 8.],
                  [8., 4.]])

Y = torch.linalg.solve(A, -C)
X = torch.linalg.solve(B.T, Y.T).T

print("X =")
print(X)
print("Проверка A @ X @ B =")
print(A @ X @ B)
````

1.4.8 Проверить, является ли система векторов a1 = (3; −3; 0; 7),
a2 = (2; 2; 4; 7), a3 = (1; 2; 3; 4), a4 = (5; −4; 1; 3) линейно зависимой?

````python
vectors = torch.tensor([[3.,  2., 1.,  5.],
                        [-3., 2., 2., -4.],
                        [0.,  4., 3.,  1.],
                        [7.,  7., 4.,  3.]], dtype=torch.float64)

rank = torch.linalg.matrix_rank(vectors)
print("Ранг системы:", rank.item())
print("Число векторов:", vectors.shape[1])
print("Линейно зависима:", rank < vectors.shape[1])
````
